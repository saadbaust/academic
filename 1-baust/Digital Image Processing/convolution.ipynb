# Create Image and Add Noise

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

# Create a synthetic image (white square + red circle on black background)
image = np.zeros((256, 256, 3), dtype=np.uint8)
cv2.rectangle(image, (30, 30), (220, 220), (255, 255, 255), -1)
cv2.circle(image, (128, 128), 50, (0, 0, 255), -1)
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)  # convert to RGB for matplotlib

# Add Gaussian noise
noise = np.random.normal(0, 80, image.shape).astype(np.float32)
noisy_image = np.clip(image.astype(np.float32) + noise, 0, 255).astype(np.uint8)

# Show original and noisy images
plt.figure(figsize=(8,4))
plt.subplot(1,2,1)
plt.imshow(image)
plt.title('Original')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(noisy_image)
plt.title('Noisy')
plt.axis('off')

plt.tight_layout()
plt.show()
```

# 1. Identity Kernel

```python
# Identity kernel
identity_kernel = np.array([[0,0,0],
                            [0,1,0],
                            [0,0,0]], dtype=np.float32)

# Apply filter2D (convolution)
identity_result = cv2.filter2D(image, -1, identity_kernel)

# Show side by side: original noisy image & filtered image
plt.figure(figsize=(8,4))
plt.subplot(1,2,1)
plt.imshow(image)
plt.title('Original')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(identity_result)
plt.title('Identity')
plt.axis('off')

plt.tight_layout()
plt.show()
```

# 2. Box Blur Kernel

```python
# Box Blur kernel
box_kernel = np.array([[1/9, 1/9, 1/9],
                       [1/9, 1/9, 1/9],
                       [1/9, 1/9, 1/9]], np.float32)

# Apply kernel
box_result = cv2.filter2D(noisy_image, -1, box_kernel)

# Show side by side: original noisy image & filtered image
plt.figure(figsize=(8,4))
plt.subplot(1,2,1)
plt.imshow(noisy_image)
plt.title('Original Noisy')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(box_result)
plt.title('Box Blur')
plt.axis('off')

plt.tight_layout()
plt.show()
```

# 3. Gaussian Blur Kernel

```python
# Gaussian Blur kernel
gauss_kernel = np.array([[1/16, 2/16, 1/16],
                         [2/16, 4/16, 2/16],
                         [1/16, 2/16, 1/16]], np.float32)

# Apply kernel
gauss_result = cv2.filter2D(noisy_image, -1, gauss_kernel)

# Show side by side
plt.figure(figsize=(8,4))
plt.subplot(1,2,1)
plt.imshow(noisy_image)
plt.title('Original Noisy')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(gauss_result)
plt.title('Gaussian Blur')
plt.axis('off')

plt.tight_layout()
plt.show()
```

# 4. Sharpening Kernel

```python
sharpen_kernel = np.array([[0,-1,0],
                           [-1,5,-1],
                           [0,-1,0]], np.float32)

sharpen_result = cv2.filter2D(noisy_image, -1, sharpen_kernel)

plt.figure(figsize=(8,4))
plt.subplot(1,2,1)
plt.imshow(noisy_image)
plt.title('Original Noisy')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(sharpen_result)
plt.title('Sharpen')
plt.axis('off')

plt.tight_layout()
plt.show()
```

# 5. Sobel Kernel (Horizontal)

```python
sobelx_kernel = np.array([[-1,0,1],
                          [-2,0,2],
                          [-1,0,1]], np.float32)

sobelx_result = cv2.filter2D(image, -1, sobelx_kernel)

plt.figure(figsize=(8,4))
plt.subplot(1,2,1)
plt.imshow(image)
plt.title('Original')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(sobelx_result)
plt.title('Sobel X')
plt.axis('off')

plt.tight_layout()
plt.show()
```

# Morphological Operations

## Image Creation

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

# Create a black image
image = np.zeros((20, 20), dtype=np.uint8)

# Draw a white square in the middle
image[5:15, 5:15] = 255

# Add small white noise pixels
image[2,2] = 255
image[17,17] = 255
image[2,17] = 255
image[17,2] = 255

plt.figure(figsize=(4,4))
plt.imshow(image, cmap='gray')
plt.title('Original Image')
plt.axis('off')
plt.show()

plt.show()
```

```python
# Define structuring element
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (5,5))
```

# 1. Erosion

```python
# Apply erosion
erosion = cv2.erode(image, kernel, iterations=1)

# Show side by side
plt.figure(figsize=(8,4))
plt.subplot(1,2,1)
plt.imshow(image, cmap='gray')
plt.title('Original Grayscale')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(erosion, cmap='gray')
plt.title('Erosion')
plt.axis('off')

plt.tight_layout()
plt.show()
```

# 2. Dilation

```python
# Apply dilation
dilation = cv2.dilate(image, kernel, iterations=1)

plt.figure(figsize=(8,4))
plt.subplot(1,2,1)
plt.imshow(image, cmap='gray')
plt.title('Original Grayscale')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(dilation, cmap='gray')
plt.title('Dilation')
plt.axis('off')

plt.tight_layout()
plt.show()
```

# 3. Opening (Erosion → Dilation)

```python
opening = cv2.morphologyEx(image, cv2.MORPH_OPEN, kernel)

plt.figure(figsize=(8,4))
plt.subplot(1,2,1)
plt.imshow(image, cmap='gray')
plt.title('Original Grayscale')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(opening, cmap='gray')
plt.title('Opening')
plt.axis('off')

plt.tight_layout()
plt.show()
```

# 4. Closing (Dilation → Erosion)

```python
closing = cv2.morphologyEx(image, cv2.MORPH_CLOSE, kernel)

plt.figure(figsize=(8,4))
plt.subplot(1,2,1)
plt.imshow(image, cmap='gray')
plt.title('Original Grayscale')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(closing, cmap='gray')
plt.title('Closing')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
gradient = cv2.morphologyEx(image, cv2.MORPH_GRADIENT, kernel)

plt.figure(figsize=(8,4))
plt.subplot(1,2,1)
plt.imshow(image, cmap='gray')
plt.title('Original Grayscale')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(gradient, cmap='gray')
plt.title('Morphological Gradient')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
tophat = cv2.morphologyEx(image, cv2.MORPH_TOPHAT, kernel)

plt.figure(figsize=(8,4))
plt.subplot(1,2,1)
plt.imshow(image, cmap='gray')
plt.title('Original Grayscale')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(tophat, cmap='gray')
plt.title('Top Hat')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
blackhat = cv2.morphologyEx(image, cv2.MORPH_BLACKHAT, kernel)

plt.figure(figsize=(8,4))
plt.subplot(1,2,1)
plt.imshow(image, cmap='gray')
plt.title('Original Grayscale')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(blackhat, cmap='gray')
plt.title('Black Hat')
plt.axis('off')

plt.tight_layout()
plt.show()
```
